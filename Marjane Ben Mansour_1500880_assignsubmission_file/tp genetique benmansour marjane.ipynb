```python
import numpy as np
import matplotlib.pyplot as plt

T = 300 
dt = 0.1 
N_steps = int(T / dt) 

rmax = 0.10 
K = 50 
m0 = 0.07 
C0 = 3 

initial_glucose = 400 
glucose_inflow = 0 

C = np.zeros(N_steps) 
glucose_level = np.zeros(N_steps)
growth_rate = np.zeros(N_steps)
mortality_rate = np.zeros(N_steps) 

C[0] = C0
glucose_level[0] = initial_glucose
growth_rate[0] = rmax * glucose_level[0] / (K + glucose_level[0])
mortality_rate[0] = m0

for i in range(1, N_steps):
    r = rmax * glucose_level[i-1] / (K + glucose_level[i-1])
    C[i] = C[i-1] + dt * (r -m0) * C[i-1]
    glucose_level[i] = glucose_level[i-1] -dt * r * C[i-1] + glucose_inflow * dt
    growth_rate[i] = r
    mortality_rate[i] = m0
    
time = np.linspace(0, T, N_steps)
plt.figure(figsize=(12, 8))
plt.subplot(4, 1, 1)
plt.plot(time, C, label='Nombre de bactéries', color='blue')
plt.title('Concentration des bactéries -Modèle de Monod')
plt.ylabel('10³ cellules/mL')
plt.grid(True)
plt.subplot(4, 1, 2)
plt.plot(time, glucose_level, label='Concentration de glucose', color='green')
plt.title('Concentration de glucose')
plt.ylabel('g/L')
plt.grid(True) 
plt.subplot(4, 1, 3)
plt.plot(time, growth_rate, label='Taux de croissance (r)', color='red')
plt.plot(time, mortality_rate, label='Taux de mortalité (m)', color='black', linestyle='--') 
plt.title('Taux de croissance et de mortalité')
plt.ylabel('h⁻¹')
plt.legend() 
plt.grid(True)
plt.tight_layout()
plt.show()
```

La courbe bleu correspond a la croissance des bactéries qui augmente puis le glucose diminue la croissance ralentit . Finalement la population peut diminue peut diminuer si la mortalité devient superieure à la croissance. Le glucose diminue progressivement car il est consommé. Le glucose diminue avec la baisse du glucose et le taux de mmortalité reste constant.

Le facteur limitant de cette expérience est le glucose. Quand il devient faible le taux de croissance diminue et la population ne peut plus augmenter .

La population décroit lorsque r est inférieur a m , la mortalité devient alors supérieure à la croissance.


En faisant varier les paramètres , normalement plus rmax sera élevé plus la croissance sera rapide .
Plus m0 sera élevé plus il y a une diminution de la population.
Si K est petit , il yaura une meilleure affinité pour le glucose , et si K est grand la croissance sera plus difficile lorsque le glucose est faible

Pour que la population se maintienne dans le temps , il faut que il yai un apport continu en glucose et une croissance suffisante ainsi quune mortalité faible.

```python
import numpy as np
import matplotlib.pyplot as plt

T = 2000 
dt = 0.1 
N_steps = int(T / dt) 

rmax = 0.10 
K = 50 
m0 = 0.07 
C0 = 10 

initial_glucose = 140 
glucose_inflow = 3 

C = np.zeros(N_steps) 
glucose_level = np.zeros(N_steps)
growth_rate = np.zeros(N_steps)
mortality_rate = np.zeros(N_steps) 

C[0] = C0
glucose_level[0] = initial_glucose
growth_rate[0] = rmax * glucose_level[0] / (K + glucose_level[0])
mortality_rate[0] = m0

for i in range(1, N_steps):
    r = rmax * glucose_level[i-1] / (K + glucose_level[i-1])
    C[i] = C[i-1] + dt * (r -m0) * C[i-1]
    glucose_level[i] = glucose_level[i-1] -dt * r * C[i-1] + glucose_inflow * dt
    growth_rate[i] = r
    mortality_rate[i] = m0
    
time = np.linspace(0, T, N_steps)
plt.figure(figsize=(12, 8))
plt.subplot(4, 1, 1)
plt.plot(time, C, label='Nombre de bactéries', color='blue')
plt.title('Concentration des bactéries -Modèle de Monod')
plt.ylabel('10³ cellules/mL')
plt.grid(True)
plt.subplot(4, 1, 2)
plt.plot(time, glucose_level, label='Concentration de glucose', color='green')
plt.title('Concentration de glucose')
plt.ylabel('g/L')
plt.grid(True) 
plt.subplot(4, 1, 3)
plt.plot(time, growth_rate, label='Taux de croissance (r)', color='red')
plt.plot(time, mortality_rate, label='Taux de mortalité (m)', color='black', linestyle='--') 
plt.title('Taux de croissance et de mortalité')
plt.ylabel('h⁻¹')
plt.legend() 
plt.grid(True)
plt.tight_layout()
plt.show()
```

Au debut nous pouvons voir que la population augmente rapidement et le glucose diminue fortement. Ensuite le glucose apporté permet à la population de survivre , des oscillations apparaissent et la population finit par se stabiliser autour d'un équilibre.

La population bactérienne présente des fluctuations marquées avant de se stabiliser car les bacteries consomment le glucose plus vite qu'il n'arrive , le glucose baisse , la croissance ralentit , la population diminue , le glucose remonte et la croissance repart. Cela crée donc des oscillations avant l'équilibre

```python
import numpy as np
import matplotlib.pyplot as plt

rmax_1 = 0.15 
m0_1 = 0.06 
K_1 = 800

rmax_2 = 0.15 
m0_2 = 0.06 
K_2 = 650

initial_glucose = 1200 
glucose_inflow = 4  

C0_1 = 10 
C0_2 = 10  
dt = 0.1 
T = 2000 
N_steps = int(T / dt)

C1 = np.zeros(N_steps)
C2 = np.zeros(N_steps)
glucose_level = np.zeros(N_steps)
growth_rate_1 = np.zeros(N_steps)
growth_rate_2 = np.zeros(N_steps)
mortality_rate_1 = np.zeros(N_steps)
mortality_rate_2 = np.zeros(N_steps)

C1[0] = C0_1
C2[0] = C0_2
glucose_level[0] = initial_glucose
growth_rate_1[0] = rmax_1 * glucose_level[0] / (K_1 + glucose_level[0])
growth_rate_2[0] = rmax_2 * glucose_level[0] / (K_2 + glucose_level[0])
mortality_rate_1[0] = m0_1
mortality_rate_2[0] = m0_2

for i in range(1, N_steps):
    r_1 = rmax_1 *glucose_level[i-1] / (K_1 + glucose_level[i-1])
    r_2 = rmax_2 * glucose_level[i-1] / (K_2 + glucose_level[i-1])
    
    m_1 = m0_1 + 0.01 * (1 -glucose_level[i-1] / initial_glucose)
    m_2 =m0_2 + 0.01 * (1 -glucose_level[i-1] / initial_glucose)
    
    m_1 = m0_1 
    m_2 = m0_2
    
    C1[i] = C1[i-1] + dt * (r_1 -m_1) * C1[i-1]
    C2[i] = C2[i-1] + dt * (r_2 -m_2) * C2[i-1]
    
    glucose_consumption = (r_1 * C1[i-1] + r_2 * C2[i-1])
    glucose_level[i] = glucose_level[i-1] -dt * glucose_consumption+ glucose_inflow * dt
    growth_rate_1[i] = r_1
    growth_rate_2[i] = r_2
    mortality_rate_1[i] = m_1
    mortality_rate_2[i] = m_2
    
time = np.linspace(0, T, N_steps)
    
plt.figure(figsize=(12, 12))
plt.subplot(4, 1, 1)
plt.plot(time, C1, label='Souche 1', color='blue')
plt.plot(time, C2, label='Souche 2', color='orange')
plt.title('Compétition entre 2 souches -Modèle de Monod')
plt.ylabel('10³ cellules/mL')
plt.legend()
plt.grid(True)

plt.subplot(4, 1, 2)
plt.plot(time, glucose_level, label='Glucose', color='green')
plt.title('Concentration de glucose')
plt.ylabel('g/L')
plt.grid(True)

plt.subplot(4, 1, 3)
plt.plot(time, growth_rate_1, label='Taux de croissance (Souche 1)', color='blue', linestyle='--')
plt.plot(time, growth_rate_2, label='Taux de croissance (Souche 2)', color='orange', linestyle='--')
plt.plot(time, mortality_rate_1, label='Taux de mortalité (Souche 1)', color='blue', linestyle=':')
plt.plot(time, mortality_rate_2, label='Taux de mortalité (Souche 2)', color='orange', linestyle=':')
plt.title('Taux de croissance (r) et de mortalité (m)')
plt.ylabel('h⁻¹')
plt.legend()
plt.grid(True)

plt.subplot(4, 1, 4)
plt.plot(time, mortality_rate_1, label='Taux de mortalité (Souche 1)', color='blue', linestyle=':')
plt.plot(time, mortality_rate_2, label='Taux de mortalité (Souche 2)', color='orange', linestyle=':')
plt.title('Taux de mortalité (m) au fil du temps')
plt.xlabel('Temps')
plt.ylabel('Taux de mortalité (m)')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

Les courbes montrent que les deux souches se developpent au debut , mais la souche 2 possède un meilleur K, elle utilise le glucose plus facilement et elle finit paar dominer la souche 1.

```python
import numpy as np
import matplotlib.pyplot as plt

rmax_1 = 0.15 
m0_1 = 0.07
K_1 = 80
resistance_1 = 0

rmax_2 = 0.15
m0_2 = 0.07
K_2 = 650
resistance_2 = 0 

initial_glucose = 1200
glucose_inflow = 1

initial_antibiotic = 0
antibiotic_intensity = 3 
antibiotic_decay = 0.03
antibiotic_inflow = 0 
antibiotic_start_time = 50

C0_1 = 10 
C0_2 = 10

dt = 0.1 
T = 200
N_steps = int(T / dt)

C1 = np.zeros(N_steps)
C2 = np.zeros(N_steps)
glucose_level = np.zeros(N_steps)
antibiotic_level = np.zeros(N_steps)
growth_rate_1 = np.zeros(N_steps)
growth_rate_2 = np.zeros(N_steps)
mortality_rate_1 = np.zeros(N_steps)
mortality_rate_2 = np.zeros(N_steps)

C1[0] = C0_1
C2[0] = C0_2
glucose_level[0] = initial_glucose
antibiotic_level[0] = 0
growth_rate_1[0] = rmax_1 * glucose_level[0] / (K_1 + glucose_level[0])
growth_rate_2[0] = rmax_2 * glucose_level[0] / (K_2 + glucose_level[0])
mortality_rate_1[0] = m0_1
mortality_rate_2[0] = m0_2

for i in range(1, N_steps):
    current_time = i * dt
    
    r_1 = rmax_1 * glucose_level[i-1] / (K_1 + glucose_level[i-1])
    r_2 = rmax_2 * glucose_level[i-1] / (K_2 + glucose_level[i-1])
    
    if current_time < antibiotic_start_time:
        antibiotic_level[i] = 0
    else:
        if antibiotic_level[i-1] == 0: 
            antibiotic_level[i] = initial_antibiotic
        else:
            antibiotic_level[i] = antibiotic_level[i-1] * (1 -antibiotic_decay * dt) + antibiotic_inflow* dt
                
    if antibiotic_level[i] > 0:
        antibiotic_effect_1 = antibiotic_intensity * (antibiotic_level[i] / initial_antibiotic) * (1 -resistance_1)
        antibiotic_effect_2 = antibiotic_intensity * (antibiotic_level[i] / initial_antibiotic) * (1 -resistance_2)
    else:
        antibiotic_effect_1 = 0
        antibiotic_effect_2 = 0
                            
    m_1 = m0_1 * (1 + antibiotic_effect_1)
    m_2 = m0_2 * (1 + antibiotic_effect_2)

    C1[i] = C1[i-1] + dt * (r_1 -m_1) * C1[i-1]
    C2[i] = C2[i-1] + dt * (r_2 -m_2) * C2[i-1]
    
# Consommation de glucose
    glucose_consumption = (r_1 * C1[i-1] + r_2 * C2[i-1])
    glucose_level[i] = glucose_level[i-1] -dt * glucose_consumption + glucose_inflow * dt
# Stockage des taux
    growth_rate_1[i] = r_1
    growth_rate_2[i] = r_2
    mortality_rate_1[i] = m_1
    mortality_rate_2[i] = m_2
    # =============================================# TEMPS POUR L'AXE DES ABCISSES# =============================================
time = np.linspace(0, T, N_steps)# =============================================# CALCUL DES ÉCHELLES COMMUNES POUR LES TAUX# =============================================# Trouver le maximum global pour les taux de croissance et mortalité
max_growth = max(np.max(growth_rate_1), np.max(growth_rate_2))
max_mortality = max(np.max(mortality_rate_1), np.max(mortality_rate_2))
y_max_rates = max(max_growth, max_mortality) * 1.1 

plt.figure(figsize=(12, 15))
           
plt.subplot(5, 1, 1)
plt.plot(time, C1, label=f'Souche 1 (Résistance = {resistance_1})', color='blue')
plt.plot(time, C2, label=f'Souche 2 (Résistance = {resistance_2})', color='orange')
plt.axvline(x=antibiotic_start_time, color='red', linestyle='--', label=f'Injection antibiotique (t={antibiotic_start_time}h)')
plt.title('Compétition entre 2 souches avec antibiotique injecté à t=500h')
plt.ylabel('10³ cellules/mL')
plt.legend()
plt.grid(True)
plt.ylim(bottom=0)
           
plt.subplot(5, 1, 2)
plt.plot(time, glucose_level, label='Glucose', color='green')
plt.title('Concentration de glucose')
plt.ylabel('g/L')
plt.legend()
plt.grid(True)
plt.ylim(bottom=0)# 3. Concentration d'antibiotique

plt.subplot(5, 1, 3)
plt.plot(time, antibiotic_level, label='Antibiotique', color='red')
plt.axvline(x=antibiotic_start_time, color='black', linestyle=':', label=f'Début injection (t={antibiotic_start_time}h)')
plt.title(f'Concentration d\'antibiotique (Décroissance = {antibiotic_decay} h⁻¹, Inflow = {antibiotic_inflow}mg/L/h)')
plt.ylabel('mg/L')
plt.legend()
plt.grid(True)
plt.ylim(bottom=0)
plt.subplot(5, 1, 4)
plt.plot(time, growth_rate_1, label='Taux de croissance (Souche 1)', color='blue', linestyle='--')
plt.plot(time, mortality_rate_1, label='Taux de mortalité (Souche 1)', color='blue', linestyle=':')
plt.title('Taux de croissance et mortalité -Souche 1')
plt.ylabel('h⁻¹')
plt.legend()
plt.grid(True)
plt.ylim(bottom=0, top=y_max_rates) 

plt.subplot(5, 1, 5)
plt.plot(time, growth_rate_2, label='Taux de croissance (Souche 2)', color='orange', linestyle='--')
plt.plot(time, mortality_rate_2, label='Taux de mortalité (Souche 2)', color='orange', linestyle=':')
plt.title('Taux de croissance et mortalité -Souche 2')
plt.xlabel('Temps (heures)')
plt.ylabel('h⁻¹')
plt.legend()
plt.grid(True)
plt.ylim(bottom=0, top=y_max_rates)

plt.tight_layout()
plt.show()
```
