Lorsque rmax est négatif, la croissance bactérienne diminue progressivement de façon exponentielle.

Le modèle de Malthus suppose que la croissance est illimitée, ce qui pose problème car en réalité les ressources sont limitées ralentissant et stabilisant la population bactérienne.
Pour le rendre plus réaliste, il faudrait donc introduire des ressources limitées, des possibles compétitions entre individus ou encore des mutations provoquant des maladies ...

```python
import numpy as np
import matplotlib.pyplot as plt
T=300
dt=0.1
N_steps = int(T / dt)
rmax=0.10
K=50
m0=0.07
C0=3
initial_glucose = 400
glucose_inflow = 0
C = np.zeros(N_steps)
glucose_level = np.zeros(N_steps)
growth_rate = np.zeros(N_steps)
mortality_rate = np.zeros(N_steps)
C[0] = C0
glucose_level[0] = initial_glucose
growth_rate[0] = rmax * glucose_level[0] / (K + glucose_level[0])
mortality_rate[0] = m0
for i in range(1, N_steps):
    r = rmax * glucose_level[i-1] / (K + glucose_level[i-1])
    C[i] = C[i-1] + dt * (r -m0) * C[i-1]
    glucose_level[i] = glucose_level[i-1] -dt * r * C[i-1] + glucose_inflow * dt
    growth_rate[i] = r
    mortality_rate[i] = m0
time = np.linspace(0, T, N_steps)
plt.figure(figsize=(12, 8))
plt.subplot(4, 1, 1)
plt.plot(time, C, label='Nombre de bactéries', color='blue')
plt.title('Concentration des bactéries -Modèle de Monod')
plt.ylabel('10³ cellules/mL')
plt.grid(True)
plt.subplot(4, 1, 2)
plt.plot(time, glucose_level, label='Concentration de glucose', color='green')
plt.title('Concentration de glucose')
plt.ylabel('g/L')
plt.grid(True)
plt.subplot(4, 1, 3)
plt.plot(time, growth_rate, label='Taux de croissance (r)', color='red')
plt.plot(time, mortality_rate, label='Taux de mortalité (m)', color='black', linestyle='--')
plt.title('Taux de croissance et de mortalité')
plt.ylabel('h⁻¹')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

Dans le 1er graphique, la population bactérienne augmente au début grâce au glucose et diminue au fur et à mesure que la concentreation en glucose diminue.
Dans le 2ème graphique, la concentration du glucose diminue progressivement au cours du temps car il est consommé par les bactéries.
Dans le 3ème graphique, le taux de croissance est élevé au début lorsque le glucose est abondant mais diminue progressivement avec la diminution en glucose.
Le taux de mortalité reste cependant constant car il a été fixé au début du programme.

Dans le 1er graphique,la population bactérienne augmente fortement au début atteignant un maximum à 90 x 10³ cellules/mL. Puis elle diminue lorsque les ressources deviennent limitées avec plusieurs oscillations pour enfin se stabiliser vers 40 x 10³ cellules/mL.
Dans le 2ème graphique,la concentration du glucose varie également en oscillations. Lorsque la population bactérienne augmente, le glucose diminue  du à sa consommation par les bactéries.
Dans le 3ème graphique, au début,le taux de croissance est supérieur au taux de mortalité, ce qui entraine une augmentation de la population. Lorsque le taux de croissance devient inférieur au taux de mortalité, la population diminue.Et enfin les 2 taux finissent par se stabilisés et devenir égaux, ce qui traduit un état d'équilibre.

Les fluctuations sont dues aux interactions entre le glucose et les bactéries. Lorsque le glucose est abondant, les bactéries se multiplient rapidement et donc croîent du à leur consommation du glucose. Et inversement lorsque la quantité en glucose diminue, la croissance diminue jusqu'à atteindre un équilibre.

```python
import numpy as np
import matplotlib.pyplot as plt
rmax_1 = 0.15
m0_1 = 0.06
K_1 = 800
rmax_2 = 0.15
m0_2 = 0.06
K_2 = 650
initial_glucose = 1200 
glucose_inflow = 4 
C0_1 = 10 
C0_2 = 10 
dt = 0.1
T = 2000 
N_steps = int(T / dt)
C1 = np.zeros(N_steps)
C2 = np.zeros(N_steps)
glucose_level = np.zeros(N_steps)
growth_rate_1 = np.zeros(N_steps)
growth_rate_2 = np.zeros(N_steps)
mortality_rate_1 = np.zeros(N_steps)
mortality_rate_2 = np.zeros(N_steps)
C1[0] = C0_1
C2[0] = C0_2
glucose_level[0] = initial_glucose
growth_rate_1[0] = rmax_1 * glucose_level[0] / (K_1 + glucose_level[0])
growth_rate_2[0] = rmax_2 * glucose_level[0] / (K_2 + glucose_level[0])
mortality_rate_1[0] = m0_1
mortality_rate_2[0] = m0_2
for i in range(1, N_steps):
    r_1 = rmax_1 *glucose_level[i-1] / (K_1 + glucose_level[i-1])
    r_2 = rmax_2 * glucose_level[i-1] / (K_2 + glucose_level[i-1])
    m_1 = m0_1
    m_2 = m0_2
    C1[i] = C1[i-1] + dt * (r_1 -m_1) * C1[i-1]
    C2[i] = C2[i-1] + dt * (r_2 -m_2) * C2[i-1]
    glucose_consumption = (r_1 * C1[i-1] + r_2 * C2[i-1])
    glucose_level[i] = glucose_level[i-1] -dt * glucose_consumption+ glucose_inflow * dt
    growth_rate_1[i] = r_1
    growth_rate_2[i] = r_2
    mortality_rate_1[i] = m_1
    mortality_rate_2[i] = m_2
time = np.linspace(0, T, N_steps)
plt.figure(figsize=(12, 12))
plt.subplot(4, 1, 1)
plt.plot(time, C1, label='Souche 1', color='blue')
plt.plot(time, C2, label='Souche 2', color='orange')
plt.title('Compétition entre 2 souches -Modèle de Monod')
plt.ylabel('10³ cellules/mL')
plt.legend()
plt.grid(True)
plt.subplot(4, 1, 2)
plt.plot(time, glucose_level, label='Glucose', color='green')
plt.title('Concentration de glucose')
plt.ylabel('g/L')
plt.grid(True)
plt.subplot(4, 1, 3)
plt.plot(time, growth_rate_1, label='Taux de croissance (Souche 1)', color='blue', linestyle='--')
plt.plot(time, growth_rate_2, label='Taux de croissance (Souche 2)', color='orange', linestyle='--')
plt.plot(time, mortality_rate_1, label='Taux de mortalité (Souche 1)', color='blue', linestyle=':')
plt.plot(time, mortality_rate_2, label='Taux de mortalité (Souche 2)', color='orange', linestyle=':')
plt.title('Taux de croissance (r) et de mortalité (m)')
plt.ylabel('h⁻¹')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

Les 2 souches présentent une forte croissance suivie d'oscillations. La souche 2 atteint cependant plus rapidement une grande population et reste plus élevée tout le long par rapport à la souche 1.
La concentration de glucose diminue fortement au début dû à sa consommation par les 2 souches puis augmente ensuite jusqu'à atteindre un équilibre.
Le taux de croissance ainsi que celui de mortalité des 2 souches restent constants au cours du temps 

La souche 2 domine la souche 1 en atteignant une population plus élevée. Cela s'explique possiblement par une meilleure affinité de celle-ci au glucose. Donc cela entraine une suivie supérieure pour la souche 2 tandis que la souche 1 est éliminée.

```python
import numpy as np
import matplotlib.pyplot as plt
rmax_1 = 0.15
m0_1 = 0.07
K_1 = 800
resistance_1 = 0
rmax_2 = 0.15
m0_2 = 0.07
K_2 = 650
resistance_2 = 0
initial_glucose = 1200
glucose_inflow = 1 
initial_antibiotic = 0
antibiotic_intensity = 3 
antibiotic_decay = 0.03
antibiotic_inflow = 0.05
antibiotic_start_time = 50 
C0_1 = 10 
C0_2 = 10 
dt = 0.1
T = 200
N_steps = int(T / dt)
C1 = np.zeros(N_steps)
C2 = np.zeros(N_steps)
glucose_level = np.zeros(N_steps)
antibiotic_level = np.zeros(N_steps)
growth_rate_1 = np.zeros(N_steps)
growth_rate_2 = np.zeros(N_steps)
mortality_rate_1 = np.zeros(N_steps)
mortality_rate_2 = np.zeros(N_steps)
C1[0] = C0_1
C2[0] = C0_2
glucose_level[0] = initial_glucose
antibiotic_level[0] = 0
growth_rate_1[0] = rmax_1 * glucose_level[0] / (K_1 + glucose_level[0])
growth_rate_2[0] = rmax_2 * glucose_level[0] / (K_2 + glucose_level[0])
mortality_rate_1[0] = m0_1
mortality_rate_2[0] = m0_2
for i in range(1, N_steps):
    current_time = i * dt
    r_1 = rmax_1 * glucose_level[i-1] / (K_1 + glucose_level[i-1])
    r_2 = rmax_2 * glucose_level[i-1] / (K_2 + glucose_level[i-1])
    if current_time < antibiotic_start_time:
        antibiotic_level[i] = 0
    else:
        if antibiotic_level[i-1] == 0:
           antibiotic_level[i] = initial_antibiotic
        else:
            antibiotic_level[i] = antibiotic_level[i-1] * (1 -antibiotic_decay * dt) + antibiotic_inflow* dt
            
    if antibiotic_level[i] > 0:
        antibiotic_effect_1 = antibiotic_intensity * (antibiotic_level[i] / initial_antibiotic) * (1 -resistance_1)
        antibiotic_effect_2 = antibiotic_intensity * (antibiotic_level[i] / initial_antibiotic) * (1 -resistance_2)
    else:
        antibiotic_effect_1 = 0
        antibiotic_effect_2 = 0
    m_1 = m0_1 * (1 + antibiotic_effect_1)
    m_2 = m0_2 * (1 + antibiotic_effect_2)
    C1[i] = C1[i-1] + dt * (r_1 -m_1) * C1[i-1]
    C2[i] = C2[i-1] + dt * (r_2 -m_2) * C2[i-1]
    glucose_consumption = (r_1 * C1[i-1] + r_2 * C2[i-1])
    glucose_level[i] = glucose_level[i-1] -dt * glucose_consumption + glucose_inflow * dt
    growth_rate_1[i] = r_1
    growth_rate_2[i] = r_2
    mortality_rate_1[i] = m_1
    mortality_rate_2[i] = m_2
time = np.linspace(0, T, N_steps)
max_growth = max(np.max(growth_rate_1), np.max(growth_rate_2))
max_mortality = max(np.max(mortality_rate_1), np.max(mortality_rate_2))
y_max_rates = max(max_growth, max_mortality) * 1.1 # +10% de marge
plt.figure(figsize=(12, 15)) 
plt.subplot(5, 1,1)
plt.plot(time, C1, label=f'Souche 1 (Résistance = {resistance_1})', color='blue')
plt.plot(time, C2, label=f'Souche 2 (Résistance = {resistance_2})', color='orange')
plt.axvline(x=antibiotic_start_time, color='red', linestyle='--', label=f'Injection antibiotique (t={antibiotic_start_time}h)')
plt.title('Compétition entre 2 souches avec antibiotique injecté à t=500h')
plt.ylabel('10³ cellules/mL')
plt.legend()
plt.grid(True)
plt.ylim(bottom=0)
plt.subplot(5, 1, 2)
plt.plot(time, glucose_level, label='Glucose', color='green')
plt.title('Concentration de glucose')
plt.ylabel('g/L')
plt.legend()
plt.grid(True)
plt.ylim(bottom=0)
plt.subplot(5, 1, 3)
plt.plot(time, antibiotic_level, label='Antibiotique', color='red')
plt.axvline(x=antibiotic_start_time, color='black', linestyle=':', label=f'Début injection (t={antibiotic_start_time}h)')
plt.title(f'Concentration d\'antibiotique (Décroissance = {antibiotic_decay} h⁻¹, Inflow = {antibiotic_inflow}mg/L/h)')
plt.ylabel('mg/L')
plt.legend()
plt.grid(True)
plt.ylim(bottom=0)
plt.subplot(5, 1, 4)
plt.plot(time, growth_rate_1, label='Taux de croissance (Souche 1)', color='blue', linestyle='--')
plt.plot(time, mortality_rate_1, label='Taux de mortalité (Souche 1)', color='blue', linestyle=':')
plt.title('Taux de croissance et mortalité -Souche 1')
plt.ylabel('h⁻¹')
plt.legend()
plt.grid(True)
plt.ylim(bottom=0, top=y_max_rates) 
plt.subplot(5, 1, 5)
plt.plot(time, growth_rate_2, label='Taux de croissance (Souche 2)', color='orange', linestyle='--')
plt.plot(time, mortality_rate_2, label='Taux de mortalité (Souche 2)', color='orange', linestyle=':')
plt.title('Taux de croissance et mortalité -Souche 2')
plt.xlabel('Temps (heures)')
plt.ylabel('h⁻¹')
plt.legend()
plt.grid(True)
plt.ylim(bottom=0, top=y_max_rates)
plt.tight_layout()
plt.show()




```

Sans antibiotique, les 2 souches sont en compétition pour le glucose. La souche 2 se développe davantage que la 1.

Avec une résistance_1 = 0.8, la souche 1 devient plus résistante à l'antibiotique donc sa mortalité augmente moins que celle de la souche 2.
Avec T=2000, on observe une meilleure évolution à long terme. La souche sensible diminue jusqu'à disparaitre tandis que l'autre se maintient.
Avec antibiotic_inflow = 0.05, cela exerce une sélection continue du à la présence constante d'antiobiotique. Cela favorise les souches résistantes.

La présence d'une faible quantité d'antiobiotique augmente la mortalité des bactéries sensibles. Les bactéries résistantes survivent mieux car elles sont moins affectées et se reproduisent alors davantage dans le temps et deviennent alors majoritaires. Donc une pollution d'antiobiotique dans un environnement favorise l'émergence des souches résistantes.
