```python
import numpy as np
import matplotlib.pyplot as plt
# Paramètres de la simulation
rmax = 0.2 # taux de croissance maximal (h⁻¹)
C0 = 0.1 # Concentration des cellules bactériennes au début de la simulation (10^3 cellules.mL-1)
T = 10 # durée totale de la simulation (heures)
dt = 0.1 # pas de temps (heures)

N_steps = int(T / dt)
# Simulation
time = np.linspace(0, T, N_steps)
C = C0 * np.exp(rmax * time)

# Tracé du graphique
plt.title('Graphique suivant la variation de la concentration des cellules bactériennes (10^3 cellules.mL-1) en fonction du temps (h⁻¹)') # Titre du graphique
plt.plot(time, C, label='Courbe suivant la population bactérienne') # Nom de la courbe
plt.xlabel('temps (h⁻¹)') # Axe des X
plt.ylabel('concentration des cellules bactériennes (10^3 cellules.mL-1)') # Axe des Y
plt.grid(True)
plt.legend()
plt.show()
```

Modèle de Monod : 

```python
import numpy as np
import matplotlib.pyplot as plt

#======================
# PARAMÈTRES DE LA SIMULATION
# ======================
T = 300 # Durée totale (heures)
dt = 0.1 # Pas de temps (heures)
N_steps = int(T / dt) # Nombre de pas de temps

# ======================
# PARAMÈTRES DE LA SOUCHE
# ======================
rmax = 0.10 # Taux de croissance maximal (h⁻¹)
K = 50 # Affinité pour le substrat (g/L)
m0 = 0.07 # Taux de mortalité minimal (h⁻¹)
C0 = 3 # Concentration initiale de bactéries (10³ cellules/mL)

# ======================
# PARAMÈTRES DU SUBSTRAT (GLUCOSE)
#======================
initial_glucose = 400 # Concentration initiale de glucose (g/L)
glucose_inflow = 0 # Flux entrant de glucose (g/L/h)

# ======================
# INITIALISATION DES TABLEAUX
# ======================
C = np.zeros(N_steps) # Concentration bactérienne
glucose_level = np.zeros(N_steps)
growth_rate = np.zeros(N_steps)
mortality_rate = np.zeros(N_steps)

# ======================
# CONDITIONS INITIALES
# ======================
C[0] = C0
glucose_level[0] = initial_glucose
growth_rate[0] = rmax * glucose_level[0] / (K + glucose_level[0])
mortality_rate[0] = m0

# ======================
# SIMULATION
# ======================
for i in range(1, N_steps):
    # Calcul des taux
    r = rmax * glucose_level[i-1] / (K + glucose_level[i-1])
    
    # Mise à jour de la population bactérienne
    C[i] = C[i-1] + dt * (r -m0) * C[i-1]
    
    # Mise à jour du glucose (consommation + flux entrant)
    glucose_level[i] = glucose_level[i-1] -dt * r * C[i-1] + glucose_inflow * dt
    
    # Stockage des taux
    growth_rate[i] = r
    mortality_rate[i] = m0

# ======================
# AXE DES X EN HEURES
# ======================
time = np.linspace(0, T, N_steps)

# ======================
# TRACÉ DES COURBES
# ======================
plt.figure(figsize=(12, 8))

# 1. Concentration bactérienne
plt.subplot(4, 1, 1)
plt.plot(time, C, label='Nombre de bactéries', color='blue')
plt.title('Concentration des bactéries -Modèle de Monod')
plt.ylabel('10³ cellules/mL')
plt.grid(True)

# 2. Concentration de glucose
plt.subplot(4, 1, 2)
plt.plot(time, glucose_level, label='Concentration de glucose', color='green')
plt.title('Concentration de glucose')
plt.ylabel('g/L')
plt.grid(True)

# 3. Taux de croissance ET mortalité (sur le même graphique)
plt.subplot(4, 1, 3)
plt.plot(time, growth_rate, label='Taux de croissance (r)', color='red')
plt.plot(time, mortality_rate, label='Taux de mortalité (m)', color='black', linestyle='--') # Ligne noire en pointillé
plt.title('Taux de croissance et de mortalité')
plt.ylabel('h⁻¹')
plt.legend() # Affiche la légende pour les deux courbes
plt.grid(True)

plt.tight_layout()
plt.show()
```

Dans le premier graphique, on voit que la concentration des bactéries augmente progressivement, atteint un maximum entre 150 et 200 heures, et ensuite diminue rapidement.

Dans le deuxièeme graphique, la variation de la courbe décrit une diminution progressive de la concentration de glucose. Les bactéries utilisent le glucose pour se multiplier et plus la population augmente, plus le glucose est consommé rapidement.

En effet, au début, la concentration bactérienne augmente rapidement car il y a une grande concentration de glucose dans le milieu. Ensuite, elle ralentit car la concentration de glucose dans le milieu diminue étant donné qu'elle est consommée par les bactéries.

En ce qui concerne le troisième graphique, le taux de croissance diminue au cours du temps. En effet, selon l'équation de Monod, quand S diminue, r diminue aussi.
Le taux de mortalité, lui, reste constant. En effet, m0 a une valeur fixe dans notre code.

Le facteur limitant est le glucose dans cette expérience.

La population décroit lorsque r<m car le nombre de bactéries mortes devient supèrieur au nombre de bactéries vivantes.

PARTIE 2 – MODEL DE HARDY - WEINBERG – Sélection et dérive génétique

```python
import numpy as np
import matplotlib.pyplot as plt

# Simulation
TAILLE_POP = 1000
GENERATIONS = 50

# Fréquence initiale de l'allèle A
freq_A_initiale = 0.1

# Valeurs sélectives (fitness)
# Cas 1 : sélection (ex : antibiotique ou stress)
fitness_selection = {"AA": 1.0,
                     "Aa": 0.9,
                     "aa": 0.7}

# Cas 2 : pas de sélection (équilibre de Hardy-Weinberg)
fitness_neutre = {"AA": 1.0,
                  "Aa": 1.0,
                  "aa": 1.0}

# Génération d'une population
def generer_population(freq_A, taille):
    freq_a = 1 - freq_A
    probabilites = [
        freq_A ** 2, # AA
        2 * freq_A * freq_a, # Aa
        freq_a ** 2 # aa
    ]
    return np.random.choice(["AA", "Aa", "aa"], size=taille, p=probabilites)
    
# Calcul de la fréquence allélique
def calcul_freq_A(population):
    nb_A = sum(genotype.count("A") for genotype in population)
    return nb_A / (2 * len(population))
    
# Application de la sélection
def appliquer_selection(population, fitness):
    poids = [fitness[g] for g in population]
    total = sum(poids)
    probabilites = [p / total for p in poids]
    return np.random.choice(population, size=len(population), p=probabilites)
    
# Simulation de l'évolution
def simuler(fitness):
    freq_A = freq_A_initiale
    frequences = [freq_A] # Initialisation de la liste
    population = generer_population(freq_A, TAILLE_POP)
    for i in range(GENERATIONS):
        population = appliquer_selection(population, fitness)
        freq_A = calcul_freq_A(population)
        population = generer_population(freq_A, TAILLE_POP)
        frequences.append(freq_A)
    return frequences

# Lancement des simulations
freq_selection = simuler(fitness_selection)
freq_neutre = simuler(fitness_neutre)

# Tracé des courbes
plt.figure()
plt.plot(freq_selection, label="Avec sélection")
plt.plot(freq_neutre, label="Sans sélection (Hardy-Weinberg)")
plt.xlabel("Génération")
plt.ylabel("Fréquence de l'allèle A")
plt.title("Évolution de la fréquence allélique")
plt.legend()
plt.grid(True)
plt.show()
```

Sans sélection (Hardy-Weinberg), la fréquence de l'allèle A restent plutôt stable. C'est l'équilibre Hardy-Weinberg. Tous les génotypes ont la même fitness, de 1.0.
Les faibles variations qu'on observe sont dues au hasard au niveau des reproductions, et donc la dérive génetique. Etant donné que la population est assez grande, de 1000 individus, l'effet de la dérive génétique est limité.
Avec sélection, je vois qu'il y a une augmentation exponentielle de la fréquence de l'allèle A. L'allèle A devient de plus en plus fréquente.
On le voit dans le code que la fitness de sélection est de 1.0 pour les individus (AA), 0.9 pour les (Aa) et 0.7 pour les (aa). Les individus (AA) survivent mieux, puis les inidividus (Aa) et les individus (aa).
Donc certains genotypes survivent mieux que d'autres avec sélection car ont une meilleure fitness.
Les génotypes contenant l'allèle A ayant une meilleure fitness, vont passer l'allèle A à leur descendance, causant l'augmenation de sa fréqunce et donc sa sélection. l'allèle A est favorisée par sélection naturelle.

Dans une petite population, les fréquences alléliques changent aléatoirement, à cause du hasard. 
Par exemple, dans une population de 10 individus, il devrait y avoir 50% d'allèles A et 50% d'allèles a. Par hasard, il se peut que plusieurs individus porteurs de A ne se reproduisent pas et que plusieurs individus porteurs de a produisennt beaucoup de descendants, donnant après quelques générations 35% d'individus porteurs de A et 65% d'individus porteurs de a. Pourtant aucun allèle n'a d'avantage. la dérive génetique se traduit par un hasard évolutif.
