```python
import numpy as np
import matplotlib.pyplot as plt

# Paramètres de la simulation
rmax = -0.1  # taux de croissance maximal (h⁻¹)
C0 = 0.1  # Concentration des cellules bactériennes au début de la simulation (10^3 cellules.mL-1)
T = 10 # durée totale de la simulation (heures)
dt = 0.1 # pas de temps (heures)
N_steps = int(T / dt)

# Simulation
time = np.linspace(0, T, N_steps)
C = C0 * np.exp(rmax * time)

# Tracé du graphique
plt.title('Graphique representant la croissance d une population bacterienne en condition optimales')  # Titre du graphique
plt.plot(time, C, label='Taux de croissance de la population') # Nom de la courbe
plt.xlabel('Temps en heure') # Axe des X
plt.ylabel('Concentration bacterienne en 10e3 cellules.mL-1') # Axe des Y
plt.grid(True)
plt.legend()
plt.show()
```

Avec un rmax negatif, la population diminue.

Ainsi avec le modele de Malthus, quand le rmax augmente le temps pour que la population double augmente, avec un rmax à 0 la population est constante. Avec un rmax negatif, la population diminue.

Les limites de ce modele sont que les conditions suppose que les limites sont illimités permettant une croissance exponetielle. De plus, il ne prends pas en compte la mort des cellules presente dans la population.

Afin d'ameliorer ce modele, toute les conditions et limitations naturelles doivent etre prise en compte.

```python
import numpy as np
import matplotlib.pyplot as plt

# ======================
# PARAMÈTRES DE LA SIMULATION
# ======================
T = 300          # Durée totale (heures)
dt = 0.1          # Pas de temps (heures)
N_steps = int(T / dt)  # Nombre de pas de temps

# ======================
# PARAMÈTRES DE LA SOUCHE
# ======================
rmax = 0.10       # Taux de croissance maximal (h⁻¹)
K = 50           # Affinité pour le substrat (g/L)
m0 = 0.07         # Taux de mortalité minimal (h⁻¹)
C0 = 3 # Concentration initiale de bactéries (10³ cellules/mL)

# ======================
# PARAMÈTRES DU SUBSTRAT (GLUCOSE)
# ======================
initial_glucose = 400  # Concentration initiale de glucose (g/L)
glucose_inflow = 0      # Flux entrant de glucose (g/L/h)

# ======================
# INITIALISATION DES TABLEAUX
# ======================
C = np.zeros(N_steps)          # Concentration bactérienne
glucose_level = np.zeros(N_steps)
growth_rate = np.zeros(N_steps)
mortality_rate = np.zeros(N_steps)

# ======================
# CONDITIONS INITIALES
# ======================
C[0] = C0
glucose_level[0] = initial_glucose
growth_rate[0] = rmax * glucose_level[0] / (K + glucose_level[0])
mortality_rate[0] = m0

# ======================
# SIMULATION
# ======================
for i in range(1, N_steps):
    # Calcul des taux
    r = rmax * glucose_level[i-1] / (K + glucose_level[i-1])

    # Mise à jour de la population bactérienne
    C[i] = C[i-1] + dt * (r - m0) * C[i-1]
      # Mise à jour du glucose (consommation + flux entrant)
    glucose_level[i] = glucose_level[i-1] - dt * r * C[i-1] + glucose_inflow * dt

    # Stockage des taux
    growth_rate[i] = r
    mortality_rate[i] = m0

# ======================
# AXE DES X EN HEURES
# ======================
time = np.linspace(0, T, N_steps)

# ======================
# TRACÉ DES COURBES
# ======================
plt.figure(figsize=(12, 8))

# 1. Concentration bactérienne
plt.subplot(4, 1, 1)
plt.plot(time, C, label='Nombre de bactéries', color='blue')
plt.title('Concentration des bactéries - Modèle de Monod')
plt.ylabel('10³ cellules/mL')
plt.grid(True)

# 2. Concentration de glucose
plt.subplot(4, 1, 2)
plt.plot(time, glucose_level, label='Concentration de glucose', color='green')
plt.title('Concentration de glucose')
plt.ylabel('g/L')
plt.grid(True)

# 3. Taux de croissance ET mortalité (sur le même graphique)
plt.subplot(4, 1, 3)
plt.plot(time, growth_rate, label='Taux de croissance (r)', color='red')
plt.plot(time, mortality_rate, label='Taux de mortalité (m)', color='black', linestyle='--')  # Ligne noire en pointillé
plt.title('Taux de croissance et de mortalité')
plt.ylabel('h⁻¹')
plt.legend()  # Affiche la légende pour les deux courbes
plt.grid(True)

plt.tight_layout()
plt.show()
```

Pour la premiere courbe representant les bacteries en fonctions du temps, nous voyons une croissance rapide et optimal à partir de 100 heures, passant de 20.10e3 cellule au bout de 100h a 38.10e3 au bout de 150 heures. Un pic de croissance apparait également au bout de 170 heures avec environ 45.10e3 cellules.Puis à 200 heures la population diminue jusqu'a atteindre 5.10e3 au bout de 300 heures.

Pour la courbe verte, nous voyons qu'au moment de l'augmentation de la population bacterienne(vu avec la premiere courbe) le niveau de glucose diminue. Et dans la fenetre de temps de 100 heures a 200 heures(fenetre dans laquelle le pic de cellules et atteint) la concentration passe de 320g/l a 90g/l. Ceci nous montrant une correlation entre le niveau de glucose dans le milieu et la proliferation bacterienne. En effet, une concentration ni trop elevée ni trop faible de glucose permet une proliferation maximal des bacteries. Cepandant passer sous le palier de 90 g/l, les bacteries ne peuvent plus proliferer et le nombre de bacteries diminue significativement.

La courbe rouge appuie cela en montrant un taux de croissance faible avec la diminution du glucose. 

Nous pouvons également supposer qu'avec un taux de glucose trop eleve, dans les premiere heures de l'experience, il y a saturation des ressources. Les cellules serait donc en quiescance ou en survie, expliquant le taux de croissance constant pendant toutes cette periode de saturation.Lorsque la concentration de glucose diminue et atteint la concentration ideal, le taux de croissance diminue car les cellules reprennent les divisions et ainsi reprenant leur cycle certaines cellules meurt, expliquant la diminution du taux de croissance.

Ainsi les facteurs limitant de l'experience pourrait etre la saturation des ressources ou au contraire l'abscence de glucose dans le milieu.
